# levels
> Map raw daily values to integer activity levels.

```python
#| default_exp levels
```

```python
#| export
import statistics
from bisect import bisect_right
```

```python
#| export
def to_level(value, thresholds):
    "Integer activity level 0..len(thresholds)+1; 0 when value<=0."
    if value <= 0: return 0
    return 1 + bisect_right(thresholds, value)

def quantile_thresholds(values, level_max):
    "Ascending cut points (len level_max-1) splitting positive `values` into `level_max` groups."
    pos = sorted(v for v in values if v > 0)
    if level_max <= 1 or len(pos) < 2: return []
    return statistics.quantiles(pos, n=level_max, method="inclusive")
```

```python
from fastcore.test import test_eq

# to_level: 0/negative -> level 0; positives bucket by bisect_right
test_eq(to_level(0, [1, 5, 10]), 0)
test_eq(to_level(-3, [1, 5, 10]), 0)
test_eq(to_level(0.5, [1, 5, 10]), 1)
test_eq(to_level(1, [1, 5, 10]), 2)
test_eq(to_level(7, [1, 5, 10]), 3)
test_eq(to_level(999, [1, 5, 10]), 4)

# quantile_thresholds: length is level_max - 1, positives only, ties allowed
test_eq(len(quantile_thresholds([1, 2, 3, 4, 5, 6, 7, 8], 4)), 3)
test_eq(quantile_thresholds([5], 4), [])              # <2 positive -> []
test_eq(quantile_thresholds([1, 2, 3], 1), [])        # level_max==1 -> []
test_eq(quantile_thresholds([0, 0, 5, 5], 4), [5.0, 5.0, 5.0])  # non-positive ignored; ties ok
```
